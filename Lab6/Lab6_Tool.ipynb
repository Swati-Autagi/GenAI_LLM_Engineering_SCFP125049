```python
@tool
def convert_marks_to_cgpa(marks:float)-> float:
  """This Function Convert Marks into Cgpa"""
  return round(marks/9.6)
```

```python
while True:
  user_query=input("Enter the user query")
  if user_query.lower()=="exit":
    break
  ai_response=llm2_with_tool.invoke(user_query)
  if ai_response.tool_calls:
    tool_args=ai_response.tool_calls[0]['args']
    tool_response=addition.invoke(tool_args)
    print(tool_response)
  else:
    print(ai_response.content)

```

```python
from langchain_core.messages import HumanMessage, AIMessage, ToolMessage
messages = []

while True:
    user_query = input("Enter user query: ")

    if user_query.lower() == "exit":
        break

    messages.append(HumanMessage(content=user_query))

    ai_response = llm2_with_tool.invoke(messages)

    messages.append(ai_response)

    if ai_response.tool_calls:
        tool_args = ai_response.tool_calls[0]["args"]

        tool_response = addition.invoke(tool_args)

        messages.append(
            ToolMessage(
                content=str(tool_response),
                tool_call_id=ai_response.tool_calls[0]["id"]
            )
        )

        response = llm2_with_tool.invoke(messages)
        messages.append(response)

        print(response.content)

    else:
        print(ai_response.content)
```
